## Find s algorithm

```python
def find_s(X,y):
  hypothesis=None
  print("initial hypothesis:",hypothesis)
  for i in range (len(X)):
    if y[i]=="Yes":
      if hypothesis is None:
        hypothesis=X[i].copy()
      else:
        for j in range(len(hypothesis)):
          if hypothesis[j]!=X[i][j]:
            hypothesis[j]="?"
    print(f"\nAfter training example{i+1}")
    print(hypothesis)
 return hypothesis
final_hypothesis=find_s(X,y)
print("\nFinal hypothesis")
```

Candidate elimination algorithm

```python
from types import GenericAlias
import copy
def candidate_elimination(concepts,target):
  speific=concepts[0].copy()
  general=[["?"for i in range(len(speific))]]
  print("\nInitial S:",speific)
  print("Initial G:",general)
  for i , h in enumerate(concepts):
    if target[i]=="Yes":
      for x in range(len(speific)):
        if h[x]!=speific[x]:
          speific[x]="?"
    else:

      for g_idx in range(len(general) - 1, -1, -1):
          g = general[g_idx]
          if all(s == '?' or s == h[j] for j, s in enumerate(g)):

              if all(h[j] == '?' or h[j] == g[j] for j in range(len(h))):
                  general.pop(g_idx)


      for x in range(len(speific)):
        if h[x]!=speific[x]:
          new_general_hyp = ["?" if j!=x else speific[x]for j in range (len(speific))]
          if new_general_hyp not in general:
              general.append(new_general_hyp)

    print("\nExample",i+1)
    print("S=",speific)
    print("G=",general)
  return speific ,general
S,G=candidate_elimination(X,y)
print("\nFinal S:",S)
print("Final G:",G)
for g in G:
  print(g)
```
